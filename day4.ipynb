# What is a function?

* A function is a block of code which only runs when it is called.
* In Python a function is defined with the `def` keyword.
* Functions can accept any number of arguments of any type.
* Functions can optionally return data as well. This data may also be of any type. You can return more than one value as well.
* Arguments can have default values as well.
* While calling functions, the user may define arguments as keyword pairs.

```python
def cars():
  print("CarA")
  print("CarB")
  print("CarC")

def Planes():
  print("PlaneA")
  print("PlaneB")
  print("PlaneC")

  
```

```python
cars()
Planes()
```

```python
def greet(name):
  print(f'Hello {name}')

greet("AJ")  
```

```python
def cars(brand,age,model):
  print(f'{brand} established in {age} and the models they made are {model}')

cars("Buggati",1909,["Contodiec","Bolide","Chiron","Tourbillion"])  
```

# Return Value

```python
def cube(x):
  return x**3

def sqrt(x):
  return x**(1/2)

n = 99

print(f'The Cube of {n} is {cube(n)}')
print(f'The Square Root of {n} is {sqrt(n)}')
```

```python
def return_func():
  return "A", "B", "C"

a,b,c = return_func()

print(f'{a} {b} {c}')
```

```python
def cars(brand,age,model):
  print(f'{brand} established in {age} and the models they made are {model}')

cars(age = 1909,brand="Buggati",model=["Contodiec","Bolide","Chiron","Tourbillion"])  
```

```python
m = 10
def calling():
  global m
  m = 56
  print(m)

print(m)
calling()
print(m)
```

```python
def square(x):
  return x**2

l = [2,3,4,5,6,7,8]

c = list(map(square,l))
print(c)
```

```python
# Lambda function are anonymous function
square2 = lambda x:x**2  # lambda argument : work
multiply = lambda a,b:a*b  # lambda argument : work
print(square2(4))
print(multiply(7,8))
```
