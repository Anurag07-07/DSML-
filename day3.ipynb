````python
for i in range(1,11,1):
  if i%3==0:
    continue   # skip the condition
  print(i,end=" ")
````

## Python Lists

* **Lists are used to store multiple items in a single variable**
* **Lists are created using square brackets**
```python
my_list = ["apple", "banana", "cherry"]

```


* **List items are ordered, changeable, and allow duplicate values.**
```python
# Duplicate values are allowed
fruits = ["apple", "banana", "apple"] 

```


* **List items are indexed, the first item has index `[0]`, the second item has index `[1]` etc.**
```python
# Indexing example
print(fruits[0])  # Output: apple
print(fruits[1])  # Output: banana

```


* **The list is changeable, meaning that we can change, add, and remove items in a list after it has been created.**
```python
# Changing an item
fruits[1] = "blueberry"

# Adding an item
fruits.append("orange")

# Removing an item
fruits.remove("apple")

```


* **You can also use the list constructor - `list()` to create lists.**
```python
# Creating a list with the constructor
another_list = list(("apple", "banana", "cherry"))

```

````python
fruits = ["apple","pineapple","rasberry","Grapes"]
print(len(fruits))

print(fruits[0])

# Add 
fruits.append("Mango")
print(fruits)

# Iterate
for f in fruits:
  print(f,end=" ")

print()

# Remove and Store
last_fruit = fruits.pop()
print(last_fruit)

# Remove 
fruits.remove("apple")
print(fruits)

# Sort 
fruits.sort()
print(fruits)

#swap
fruits[0],fruits[len(fruits)-1] = fruits[len(fruits)-1],fruits[0]
print(fruits)
````

````python
n = input("Enter the List")

# Convert String to List
n = n.split(" ")

# Convert the String into int
for i in range(0,len(n)):
  n[i] = int(n[i])

print(n)
````
