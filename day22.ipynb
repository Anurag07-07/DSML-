# Types of Plots 

```python
plt.bar([1,2,3],[10,20,5])
plt.show()
```

```python
plt.bar([1,2,3],[10,20,5],width=0.2)
plt.show()
```

```python
plt.bar([1,2,3],[10,20,5],width=0.2,color="b")
plt.legend(["Shop-1"])
plt.show()
```

```python
plt.bar([1,2,3],[10,20,5],width=0.2)
plt.bar([1,2,3],[5,10,6],width=0.2)
plt.legend(["Shop-1","Shop-2"])
plt.show()
```

```python
plt.bar([1,2,3],[10,20,5],width=0.2)
plt.bar([1.2,2.2,3.2],[5,10,6],width=0.2)
plt.legend(["Shop-1","Shop-2"])
plt.show()
```

```python
plt.bar([1,2,3],[10,20,5],width=0.2)
plt.bar(np.array([1,2,3])+0.2,[5,10,6],width=0.2)
plt.legend(["Shop-1","Shop-2"])
plt.show()
```
