```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(20)
y1 = np.linspace(1,20,20)+np.random.rand(20)*5
y2 = np.linspace(1,20,20)+np.random.rand(20)*10
plt.scatter(x,y1,s=y1*5)
plt.scatter(x,y2,s=y2*5)
plt.show()
```

```python
# Sigmoid Function or Logistic Function
import math

sigmoid = lambda x:1/(1+math.exp(-x))
x = np.arange(-100,100)
y = [sigmoid(i) for i in x]

plt.plot(x,y)

plt.show()
```

```python
# Sigmoid Function or Logistic Function
import math

sigmoid = lambda x:1/(1+math.exp(-x))
x = np.arange(-100,100)
y = [sigmoid(i) for i in x]

plt.plot(x,y)

plt.xlim(-10,10)
plt.show()
```

# Subplots

```python
x = np.arange(1,10,0.1)
y = np.sin(x)

plt.plot(x,y)
```

```python
x = np.arange(1,10,0.1)
y1 = np.sin(x)
y2 = np.tan(x)

fig,ax = plt.subplots(2,1)
ax[0].plot(x,y1)
ax[1].plot(x,y2)
plt.show()
```

```python
x = np.arange(1,10,0.1)
y1 = np.sin(x)
y2 = np.tan(x)
y3 = np.log(x)

fig,ax = plt.subplots(3,1)
ax[0].plot(x,y1)
ax[1].plot(x,y2)
ax[2].plot(x,y3)
plt.show()
```

```python
x = np.arange(1,10,0.1)
y1 = np.sin(x)
y2 = np.tan(x)
y3 = np.log(x)


plt.figure()

plt.subplot(2,3,1)
plt.plot(x,y1)
plt.subplot(2,3,3)
plt.plot(x,y2)
plt.subplot(2,3,5)
plt.plot(x,y3)

plt.show()
```

```python
a = np.arange(-1,1,0.005)
b=a 
a,b = np.meshgrid(a,b)

fig = plt.figure()
ax = fig.add_subplot(projection="3d")
ax.plot_surface(a,b,a**2+b**2) # paraboloid
plt.show()
```
