# List is Sorted or Not

```python

a = (1,4,10,12,14,18,20,20)

x = True

for i in range(0,len(a)-1,1):
  if a[i]>a[i+1]:
    x = False
    break

if x:
  print("Sorted")
else: 
  print("Not Sorted")
```

# Args and Kwargs

```python
# * represent the spread operator
def product(*args):     # we can access element same as list
  print(args,type(args),args[0],args[1],args[2])
  p = 1
  for i in args:
     p = p*i

  return p

p = product(45,56,89)
print(f'The Product of Given Numbers is {p}')
```

```python
currency = dict()
currency = {
  'India':"Rupees",
  'USA':"Dollar",
  "England":"Euros",
  (1,2,3):"ABC"
}
print(currency)
print(currency['USA'] )
print(currency[(1,2,3)])
del currency[(1,2,3)]
print(currency)
```

```python
def solve(s):
  freq= {}
  # Iterate over string
  for c in s:
    #Check if this char present or not
    if c in freq:
      freq[c]+=1
    else:
      freq[c] = 1

  odd_freq = 0

# Check how many char have odd freq
  for frequency in freq.values():
    if frequency%2==1:
      odd_freq+=1

  if frequency<=1:
    return 1
  return 0

print(solve("AAASSSDDDS"))
```
