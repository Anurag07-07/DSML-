```python
import math

print(math.comb(45,5))
print(math.perm(45,5))
# Combination
def fac(n):
  if n==0:
    return 1
  return n*fac(n-1)

def nCr(n,r):
  return fac(n)/(fac(r)*fac(n-r))


n = int(input("Enter n"))
r = int(input("Enter r"))

ans = nCr(n,r)
print(ans)
```

```python
# Permutation
def fac(n):
  if n==0:
    return 1
  return n*fac(n-1);

def nPr(n,r):
  return fac(n)/(fac(n-r))


n = int(input("Enter n"))
r = int(input("Enter r"))

ans = nPr(n,r)
print(ans)
```

```python
def decimaltobinary(n):
  pow = 1;
  ans = 0
  while n!=0:
    rem = n%2
    ans = ans+rem*pow
    pow = pow*10
    n//=2
  
  return ans

# or

def ntob(n):
  return bin(n).replace("0b","")


ans = decimaltobinary(15)
print(ans)
```

```python
def binarytodecimal(nums):
  n = int(nums)
  pow = 1
  ans = 0
  while n!=0:
    rem = n%10
    ans = ans+rem*pow
    pow = pow*2
    n//=10
  
  return ans

# or
def btod(n):
  return int(n,2)


ans = btod("1111")
print(ans)
```

```python
def trailingzeroes(n):
  i = 5
  ans = 0
  while i<=n:
    ans+= n//i
    i = i*5 
  return ans

ans = trailingzeroes(30)
print(ans)
```
