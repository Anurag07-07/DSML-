```python
class Car2:
  def __init__(self,name,speed):
    self.name = name
    self.speed = speed


  def describe(self):
    print(f'The Name is {self.name} and Speed is {self.speed}')


c2 = Car2("Koenisegg Jesko","530kmph")
c2.describe()

c1 = Car2("Yangwang U9","496kmph")
c1.describe()

c3 = Car2("Bugatti Bolide","490kmph")
c3.describe()
```

# Polymorphism

```python
class Cat:
  def sound(self):
    print("Meow")

class Dog:
  def sound(self):
    print("Woof")

def AnimalSound(animal_object):
  animal_object.sound()



c1 = Cat()
d1 = Dog()

AnimalSound(c1)
AnimalSound(d1)
```

# Encapsulation

```python
class User:
  def __init__(self,name,age):
    self._name = name # Protected
    self.__age = age # Private

  def describe(self):
    print(f'The Name is {self._name} and Age is {self.__age}')


class Employee(User):
  def __init__(self, name, age,work):
    super().__init__(name, age)
    self.work = work

u1 = User("AJ","21")
print(u1._name)
# print(u1.__age)
u1.describe()
# print(u1.__age)



# e1 = Employee("AJ","21","Something")
# print(e1.work)
```

# Abstraction

```python
from abc import ABC,abstractmethod

class Vehicle(ABC):
  @abstractmethod
  def start(self):
    pass

class Car(Vehicle):
  def start(self):
    return "Car Start..."
  
c4 = Car()
print(c4.start())
```
