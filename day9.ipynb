```python
def twoSum(arr,target):
  h = dict()
  ans = {}
  for i in range(len(arr)):
    left = target-arr[i]
    if left in h:
      return h[left],i
    if arr[i] not in h:
      h[arr[i]]=i

  return ans   

```

```python
l1 = [2,7,11,13]
ans = twoSum(l1,9)
print(list(ans))
```

```python
def maximum_abs_difference(arr):
  pref = [0]*len(arr)
  pref[len(arr)-1] = 0
  for i in range(len(arr)-2,-1,-1):
    pref[i] = max(pref[i+1],arr[i+1])
  
  ans = 0
  for i in range(len(arr)):
    diff = abs(arr[i]-pref[i])
    ans = max(diff,ans)



  print(ans)
```

```python
maximum_abs_difference([1,3,-1])
```

* given equation |a[i]-a[j]|+|i-j| 
* return max absolute difference

* Input = [1,3,-1]
* Output = 5

```python
# intution 
# think about four cases
# |a[i]| = (a[i]-a[j]) 

def maxArr(A):
  inf = 10**9
  a = -inf
  b = inf
  c = -inf
  d = inf

  for i in range(len(A)):
    x = A[i]
    a = max(a,x+i)
    b = min(b,x+i)
    c = max(c,x-i)
    d = min(c,x-i)

  ans = max(a-b,c-d)

  return ans
```

```python
ans = maxArr([1,3,-1])
print(ans)
```

Given [3,4,-1,1] Return first missing postive 
ans = 2

```python
def missing_postive(arr):
  n = len(arr)
  for i in range(len(arr)):
    while (
      arr[i]>0 and arr[i]<=n and arr[i]!=arr[arr[i]-1]):
      index = arr[i]-1
      arr[i],arr[index] = arr[index],arr[i]

  for i in range(len(arr)):
    if i+1!=arr[i]:
      return i+1
    
  return n+1
```

```python
ans = missing_postive([3,4,-1,1])
print(ans)
```
