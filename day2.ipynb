# Branching

```python
temp = int(input("Enter the Temp"))
if temp<=15:
  print(f'Cold Weather')
elif temp>15 and temp<25:
  print("Moderate Weather")  
else:
  print(f'Hot Weather')
```

# Sum of Digit 

```python
n = int(input("Enter the Number"))

sum = 0

while n!=0:
  last = n%10
  sum += last
  n = n//10

print(f'The Sum of Digit is {sum}')
```
