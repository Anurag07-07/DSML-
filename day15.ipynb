# Logical Function

```python
a = np.array([1,2,3,4])
np.any(a)
```

```python
# if there is any element in the array which is non zero it give true otherwise false

a = np.array([0,0,0,0])
np.any(a)
```

```python
# if there is any element pair where certain conditon is true
a = np.array([1,2,3,4,5])
b = np.array([1,2,6,4,8])
np.any(a<b)
```

```python
# if this condition is true for all the cases
np.all(a<b)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([1,2,6,4,8])
c = np.array([6,4,4,5,8])
np.all((a<=b) & (b<=c))
```

```python
import numpy as np

# 1. 'skiprows=1' lagane se pehli line (headers) ignore ho jayegi
# 2. 'dtype="str"' se saara data string format mein load hoga
h = np.loadtxt('steps_tracker_dataset.txt', dtype="str", skiprows=1)

# Exact list of lists format print karne ke liye .tolist() use karo
print(h)
```

```python
date,steps,distance_km,calories_burned,active_minutes,sleep_hours,water_intake_liters,mood = h.T
```

```python
print(mood)
print(sleep_hours)
print(steps)
```

```python
distance_km = np.array(distance_km).astype(float)
sleep_hours = np.array(sleep_hours).astype(float)
calories_burned = np.array(calories_burned).astype(float)
water_intake_liters = np.array(water_intake_liters).astype(float).astype(int)
active_minutes = np.array(active_minutes).astype(int)
steps = np.array(steps).astype(int)
```

```python
np.unique(mood)
print(active_minutes)
print(steps)
# tired happy stressed sad energetic 
```

```python
# energetic => e
# happy => h
# sad => s
# stressed => st
# tired => td

mood[mood == "energetic"] = "e"
mood[mood == "happy"] = "h"
mood[mood == "sad"] = "s"
mood[mood == "stressed"] = "st"
mood[mood == "tired"] = "td"
print(mood)
```

```python
active_minutes[active_minutes<100] = 0
active_minutes[active_minutes>=100] = 1

# 1 means active
# 0 means non active

print(active_minutes)
```

```python
print(f'The Average Step Count is {np.mean(steps)}')
```

```python
# On which date the step count was highest ? step_count,date
ans = date[np.argmax(steps)]
print(ans)
```
