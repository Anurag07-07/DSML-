# Higher Order Function 

```python
def higher_order_function(x): 
  return lambda n: x*n # Low Order Function
```

```python
ans = higher_order_function(5)
print(ans(7))
```

```python
list1 = [1,2,3,4,6]
print(list(map(ans,list1)))
```

```python
# Filter
print(tuple(filter(lambda x:x%2==0 ,list1)))
```

```python
#Reduce
from functools import reduce
ans = reduce(lambda x,y:x+y,list1)
print(ans)
print(list(reversed(list1)))
```

```python
inches = [1,2,3,4.5,-1,0]
valid_values = list(filter(lambda x:x>0,inches))
into_cms = map(lambda x:x*2.54,valid_values)
total_length = reduce(lambda x,y:x+y,into_cms)
print(f'The Total Length of Cloth Required {total_length}')
```
