# Measure of Despersion/Variabilty

### Mean Absolute Deviation

$$\frac{\sum_{i=1}^{m}| d_i - \text{mean}|}{m}$$

* Formula used to Calculate Variablilty

### Mean Square Error / Variance

$$\frac{\sum_{i=1}^{m}( d_i - \text{mean})^{2}}{m}$$

### Standard Deviation

$$\sqrt{\frac{\sum_{i=1}^{m}( d_i - \text{mean})^{2}}{m}}$$
