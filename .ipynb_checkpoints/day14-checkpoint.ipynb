# Working with 2D array

```python
m1 = np.array([[1,2,3],[1,2,3]])
print(m1)
print(len(m1)) # no of rows
print(m1.ndim) # know the dimension
print(m1.shape) # (no of element in row,no of element in col)
```

```python
a = np.array([1,2,3,4,5],ndmin=2)  # convert the one d array in 2 d array
print(a)
```

```python
arr = np.arange(1,13)
print(arr)
print(arr.shape)

brr = arr.reshape(3,4) # 3 row and 4 col
print(brr)
```

```python
arr = np.arange(4)
brr = np.resize(arr,(2,4))
print(brr)
```

```python
arr = np.arange(4)
arr.resize((2,4))
print(arr)
```

```python
arr = np.arange(4)
crr = arr
# arr.resize((2,4)) # this is not possible because arr refernce to crr and while referencing resize not work
arr.resize((2,4)) # use refcheck=Flase to run
print(crr)
```

```python
a = np.arange(3).reshape(1,3)
print(a)
print(a.T) # transpose
b = np.arange(3)
print(b.T) # not change no error 
```

```python
a = np.arange(12).reshape(3,4)
print(a)
```

```python
c = a.flatten() # two d to one d
print(c)
```

```python
b = a.reshape(12) # two d to one d
print(b)
```

```python
d = a.reshape(6,-1) #-1 is use to figure it out auto but for 1 dimension only 
print(d)
```

```python
# 0 matrix
z = np.zeros(5)
print(z)
# 1 matrix
o = np.ones(5)
print(o)
# 5 matrix
f = np.ones((2,5))*5 
print(f)
# diag matrix
d = np.diag([1,2,3,4,5])
print(d)
# one d array
b = np.diag(d)
print(b)
# 1 in diagonal
i = np.identity(5)
print(i)
```

```python
m1 = np.arange(1,10).reshape((3,3))
print(m1)
print(m1[1][2]) # dont use it
print(m1[1,2]) # [row,col]
```

```python
m1 = np.array([100,200,300,400,500])
print(m1[[2,3,1,4]])
```

```python
m1 = np.arange(1,10).reshape((3,3))
print(m1)
m1[[0,0,1],[1,2,1]]  #(0,1) (0,2) (1,1)
m1[0,1:3] # 0th row me 1 se 3 excluded
```
