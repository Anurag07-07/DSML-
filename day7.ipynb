```python
import math
print(math.log(1000,100)) #log(expo,base)
print(math.log2(1000)) 
print(math.log10(1000))
print(math.log1p(1000))
print(math.log1p(1000))
print(math.e)
print(math.pi)
```

```python
n = int(input())

print(math.ceil(math.log(n,2)))
```

```python
def gcd(a,b):
  while a!=0:
    a,b = b%a,a  
  return b

print(gcd(3,27))
```

```python
def prime(n):
  for i in range(2,math.floor(math.sqrt(n))+1):
    if n%i==0:
      return False
  return True

n = int(input("Enter The Number"))

print(prime(n))
```
