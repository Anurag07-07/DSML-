# Q1 Rotate Array By K Places 

```python

def reverse(nums,start,end):
  while start<=end:
    nums[start],nums[end]=nums[end],nums[start]
    start+=1
    end-=1

def rotate(nums,k):
  n = len(nums)
  k = k%n

  reverse(nums,0,k-1)
  reverse(nums,k,n-1)
  reverse(nums,0,n-1)



```

```python
l = list(map(int,input().split()))
k = int(input())

rotate(l,k)
print(l)
```

# Find Second Largest

```python
def secondlargest(nums):
  n = len(nums)
  if n<=1:return -1

  max = nums[0]
  smax = nums[1]

  for i in nums:  
    if i>max:
      smax = max
      max = i
    elif i<max and i>smax:
      smax = i

  return smax
```

```python
l = list(map(int,input().split()))

ans = secondlargest(l)
print(ans)
```

# Remove Duplicates

```python
def removeduplicates(l):
  n  = len(l)
  i = 1
  for j in range(1,n):
    if l[i-1]!=l[j]:
      l[i] = l[j]
      i+=1
  
  l = l[:i]
  return l
```

```python
l = list(map(int,input().split()))

ans = removeduplicates(l)
print(ans)
```

# List Intersection Without Set

```python
def listintersection(nums1,nums2):
  i = 0
  j = 0
  ans = []

  while i<len(nums1) and j<len(nums2):
    if nums1[i]==nums2[j]:
      ans.append(nums1[i])
      i+=1
      j+=1
    elif nums1[i]>nums2[j]:
      j+=1
    else:
      i+=1
  
  return ans
```

```python
l1 = list(map(int,input().split()))
l2 = list(map(int,input().split()))

ans = listintersection(l1,l2)
print(ans)
```

# Chuck List 

```python
def chucklist(nums):
  ans = []
  for i in range(0,len(nums),3):
    one_d = []
    for j in range(i,i+3):
      one_d.append(nums[j])
    ans.append(one_d)

  return ans
```

```python
l1 = list(map(int,input().split()))

ans = chucklist(l1)
print(ans)
```

# Running Sum 

```python
def runningsum(l1):
  size = len(l1)
  pref = [0]*size
  pref[0] = l1[0]
  for i in range(1,size):
    pref[i] = pref[i-1]+l1[i]

  return pref
```

```python
l1 = list(map(int,input().split()))

ans = runningsum(l1)
print(ans)
```

# Move Zeroes

```python
def movezereos(nums):
  i = 0
  j = 0
  while j<len(nums):
    if nums[j]!=0:
      nums[i] = nums[j]
      i+=1
    j+=1

  # Add All the zeroes in last
  while i<len(nums):
    nums[i]=0
    i+=1

  return nums  
```

```python
l1 = list(map(int,input().split()))

ans = movezereos(l1)
print(ans)
```

# Sort by Second Element

```python
def SortTheArray(nums):

  # Selection sort 
  # Find the Minimum and Swap it 

  # Finding Minimum
  for i in range(0,len(nums)-1):
    # Current Minimum
    min = i
    for j in  range(i+1,len(nums),1):
      if nums[j][1]>nums[i][1]:
        min = j
    # Swap the Values
    nums[min],nums[i] = nums[i],nums[min]

  return nums
```

```python
nums = [(1,3),(2,2),(3,1),(4,7)]
ans = SortTheArray(nums)

print(ans)
```

# Swap First and Last 

```python
def firstandlast(t):
  l = list(t)
  l[0],l[len(t)-1] = l[len(t)-1],l[0]
  t = tuple(l)
  return t
```

```python
t = (1,2,3,4)
ans = firstandlast(t)
print(ans)
```

# Find  Missing Element

```python
def FindMissingElement(set1,set2):
  ans = []
  for i in set1:
    if i not in set2:
      ans.append(i) 

  return ans 
```

```python
set1 = {1,2,3,4,5}
set2 = {1,3,5}

ans = FindMissingElement(set1,set2)
print(ans)
```

# Panagram

```python
def Panagram(s):
  new_set = set()
  for i in s:
    new_set.add(i)

  if len(new_set)<26:
    return False
  return True
```

# Synmmetric Difference

```python
def SymmetricDifference(s1,s2,s3):
 freq = dict()

 for i in s1:
  if i in freq:
   freq[i]+=1
  else:
   freq[i]=1

 for i in s2:
  if i in freq:
   freq[i]+=1
  else:
   freq[i]=1

 for i in s3:
  if i in freq:
   freq[i]+=1
  else:
   freq[i]=1

 ans = set()
 for i,j in freq.items():
  if j%2!=0:
   ans.add(i)

 return ans
   
    


```

```python
set1 = {1,2,3}
set2 = {2,4,3}
set3 = {4,5,3}


ans = SymmetricDifference(set1,set2,set3)
print(ans)
```

# Remove Duplicates Preserve Order

```python
def preserveOrder(nums):
  s = set()
  l = list()

  for i in nums:
    if i not in s:
      s.add(i)
      l.append(i)

  return l

```

```python
l1 = [4,2,3,2,1,4,5]

ans = preserveOrder(l1)
print(ans)
```

# Primes in range 

```python
def Prime(n):
  for i in range(2,n):
    if n%i==0:
      return False
    
  return True


def Range(N):
  ans = []
  for i in range(1,N+1):
    if Prime(i):
      ans.append(i)
    else:
      continue
  return ans
```

```python
n = int(input("Enter The Range"))

ans = Range(n)
print(ans)
```
