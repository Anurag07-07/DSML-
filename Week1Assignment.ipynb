# Q1 Rotate Array By K Places 

```python

def reverse(nums,start,end):
  while start<=end:
    nums[start],nums[end]=nums[end],nums[start]
    start+=1
    end-=1

def rotate(nums,k):
  n = len(nums)
  k = k%n

  reverse(nums,0,k-1)
  reverse(nums,k,n-1)
  reverse(nums,0,n-1)



```

```python
l = list(map(int,input().split()))
k = int(input())

rotate(l,k)
print(l)
```

# Find Second Largest

```python
def secondlargest(nums):
  n = len(nums)
  if n<=1:return -1

  max = nums[0]
  smax = nums[1]

  for i in nums:  
    if i>max:
      smax = max
      max = i
    elif i<max and i>smax:
      smax = i

  return smax
```

```python
l = list(map(int,input().split()))

ans = secondlargest(l)
print(ans)
```

# Remove Duplicates

```python
def removeduplicates(l):
  n  = len(l)
  i = 1
  for j in range(1,n):
    if l[i-1]!=l[j]:
      l[i] = l[j]
      i+=1
  
  l = l[:i]
  return l
```

```python
l = list(map(int,input().split()))

ans = removeduplicates(l)
print(ans)
```

# List Intersection Without Set

```python
def listintersection(nums1,nums2):
  i = 0
  j = 0
  ans = []

  while i<len(nums1) and j<len(nums2):
    if nums1[i]==nums2[j]:
      ans.append(nums1[i])
      i+=1
      j+=1
    elif nums1[i]>nums2[j]:
      j+=1
    else:
      i+=1
  
  return ans
```

```python
l1 = list(map(int,input().split()))
l2 = list(map(int,input().split()))

ans = listintersection(l1,l2)
print(ans)
```

# Chuck List 

```python
def chucklist(nums):
  ans = []
  for i in range(0,len(nums),3):
    one_d = []
    for j in range(i,i+3):
      one_d.append(nums[j])
    ans.append(one_d)

  return ans
```

```python
l1 = list(map(int,input().split()))

ans = chucklist(l1)
print(ans)
```
